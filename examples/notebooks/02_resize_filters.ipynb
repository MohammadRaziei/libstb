# Resize filters

Compare every resize filter, shrinking and enlarging. Filters are given by name; other options live on `libstb.Resizer`.

```python
import numpy as np
import matplotlib.pyplot as plt
import libstb
```

```python
def make_test_image(size=128):
    """Checkerboard over a gradient: sharp edges show off the filters."""
    y, x = np.mgrid[0:size, 0:size]
    grad = (x * 255 // (size - 1)).astype(np.uint8)
    checks = ((x // 8 + y // 8) % 2 * 255).astype(np.uint8)
    return libstb.Image(np.dstack([grad, checks, 255 - grad]))

img = make_test_image()
plt.imshow(img.numpy()); plt.axis('off'); plt.title('source 128x128');
```

## Shrinking (128 to 48)

```python
names = ["nearest", "linear", "cubic", "bspline", "mitchell", "box"]
fig, axes = plt.subplots(1, len(names), figsize=(14, 2.8))
for ax, name in zip(axes, names):
    ax.imshow(img.resize(48, 48, name).numpy(), interpolation="nearest")
    ax.set_title(name); ax.axis("off")
```

## Enlarging (128 to 384), zoomed on a corner

The top-left 96 pixels show how each filter treats the checkerboard edges.

```python
fig, axes = plt.subplots(1, len(names), figsize=(14, 2.8))
for ax, name in zip(axes, names):
    big = img.resize(384, 384, name).numpy()
    ax.imshow(big[:96, :96], interpolation="nearest")
    ax.set_title(name); ax.axis("off")
```

## Names accepted

Names are case-insensitive. An unknown name raises `ValueError` and lists the valid ones (there is no `lanczos`).

```python
try:
    img.resize(64, 64, "lanczos")
except ValueError as e:
    print(e)
```

## Other options

Edge mode and sRGB handling are keywords of `Resizer`.

```python
resizer = libstb.Resizer(libstb.Resizer.Filter.MITCHELL, libstb.Resizer.Edge.WRAP, srgb=False)
out = img.resize(64, 64, resizer)
plt.imshow(out.numpy(), interpolation="nearest"); plt.axis("off");
```
