# Resize filters

Compare every resize filter, shrinking and enlarging. Filters are given by name; other options live on `stb.Resizer`.

```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
import libstb as stb

DATA = Path.cwd().parents[1] / "data"      # data/ at the root of the repository (see data/README.md)
```

```python
# A 128 x 128 window of the astronaut photo with sharp edges (the space shuttle model): they show off the filters.
img = stb.Image.open(DATA / "images" / "astronaut.png").crop(356, 30, 128, 128)
plt.imshow(img.numpy()); plt.axis('off'); plt.title('source 128x128');
```

## Shrinking (128 to 48)

```python
names = ["nearest", "linear", "cubic", "bspline", "mitchell", "box"]
fig, axes = plt.subplots(1, len(names), figsize=(14, 2.8))
for ax, name in zip(axes, names):
    ax.imshow(img.resize(48, 48, name).numpy(), interpolation="nearest")
    ax.set_title(name); ax.axis("off")
```

## Enlarging (128 to 384), zoomed on a corner

The top-left 96 pixels show how each filter treats the edges: stair steps (nearest), blur (linear, bspline) or ringing (cubic).

```python
fig, axes = plt.subplots(1, len(names), figsize=(14, 2.8))
for ax, name in zip(axes, names):
    big = img.resize(384, 384, name).numpy()
    ax.imshow(big[:96, :96], interpolation="nearest")
    ax.set_title(name); ax.axis("off")
```

## Names accepted

Names are case-insensitive. An unknown name raises `ValueError` and lists the valid ones (there is no `lanczos`).

```python
try:
    img.resize(64, 64, "lanczos")
except ValueError as e:
    print(e)
```

## Other options

Edge mode and sRGB handling are keywords of `Resizer`.

```python
resizer = stb.Resizer(stb.Resizer.Filter.MITCHELL, stb.Resizer.Edge.WRAP, srgb=False)
out = img.resize(64, 64, resizer)
plt.imshow(out.numpy(), interpolation="nearest"); plt.axis("off");
```
